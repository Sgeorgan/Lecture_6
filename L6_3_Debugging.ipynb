# Debugging your Python code

A pivotal aspect of refining your code, once you've navigated through the syntax problems, is going through the often painful process of debugging. To judge whether your code is correct, it's crucial to have a benchmark, or a known result that the code is intended to produce. Often, this involves using simplified datasets or specific scenarios for which you can work out the answer yourself.

Let's shift our focus to a scenario involving the analysis of daylight duration at a Swedish weather station over the last decade. Without prior knowledge of the exact figures, we can't simply keep adjusting the code until it yields the anticipated variation in daylight hours. Moreover, considering we're examining a decade's worth of data, the dataset could contain thousands of entries, making it very hard to check the results by hand. Here, using a smaller subset of the data for testing becomes invaluable. For example, we might select a short segment of the dataset - perhaps the daylight duration for the first five days of January 2014 - as illustrated below (illustrative values).

```text
YEAR    MONTH DAY    SUNRISE   SUNSET    DAYLIGHT_DURATION
------  ----- ----   --------  --------  -----------------
2014    01    01     08:45     15:15     6.5 hours
2014    01    02     08:45     15:17     6.53 hours
2014    01    03     08:44     15:20     6.6 hours
2014    01    04     08:44     15:22     6.63 hours
2014    01    05     08:43     15:25     6.7 hours
```

From this snippet, we can deduce two key points:

1. The code must be able to calculate the daylight duration accurately, adjusting for the varying sunrise and sunset times, without ignoring any potential anomalies in the data (for example, missing values).
2. If our code correctly computes the daylight duration for this sample dataset - showing a gradual increase over the selected days - we can be more confident in its ability to handle the complete dataset effectively.

In cases like this, testing with a smaller, manageable subset not only simplifies the debugging process but also serves as a reliable indicator of how well the code will perform on a larger scale.

## Make your code crash quickly and regularly

This may sound silly, but it is a good thing when your code crashes the same way every time you run it. If you have different behaviors when you run your code several times without making changes to the code, it will be much more difficult to isolate the source of the problem. What we ideally want in a code is to see behavior that is **consistent**.

In addition, if you expect to debug your program efficiently, you can't afford to wait 30 minutes every time you run it in order for it to crash. If your code crashes when processing a massive data file, you can consider testing with some smaller part of the file. Does the code still crash? If not, why not? Are there some parts of the code that seem to run just fine every time? If you can reduce the time needed for a crash, and isolate where in the code the problem lies (perhaps in memory if you're dealing with really large datasets), you will save yourself time debugging.

## Make small changes and track them

It is worth noting that often we don't keep track of every single small change to our code, but rather save when things are working as we expected. This means that when you debug, you might not keep track of every little change you make. This is fine, but it is important when you are debugging that you make small changes in one part of the code, then re-test. You should change one thing at a time, test the code, and make more changes if needed. Changing several things at once might be appealing, but it will make it harder to see exactly what is causing the problem because you can't isolate the issue to a single line of the program.

> ⚠️ **Common Pitfall: a "bug" that is really just stale notebook state**
>
> In Jupyter, variables and functions stay in memory from whenever you last ran a cell - even if you have since edited or deleted that cell. This causes two very confusing situations:
>
> - Code that **works for you but fails for someone else**, because you are relying on a variable defined in a cell you have since changed.
> - Code that **still fails after you have fixed it**, because you edited a function but never re-ran its cell, so Python is still using the old version.
>
> Before spending a long time hunting a bug, always try **Kernel → Restart & Run All**. If the problem disappears, it was stale state rather than a real bug. If it remains, you now have a bug you can reproduce consistently - which, as described above, is exactly what you want when debugging.

## Summary

This lesson was about the **process** of debugging, rather than about any particular Python command. The key ideas:

- **Know what the right answer looks like.** You cannot tell whether code is correct unless you have something to compare it against - a small dataset, a hand-calculated value, or a known result.
- **Test on a small subset first.** A few rows that you can check by hand are far more useful than thousands that you cannot.
- **Make the crash consistent and fast.** A bug that appears every time, quickly, is much easier to find than one that appears occasionally after a long run.
- **Change one thing at a time, then re-test.** Changing several things at once makes it impossible to tell which change actually mattered.

Debugging is a normal, expected part of programming - not a sign that you are doing something wrong. Every programmer, however experienced, spends a good share of their time here.
