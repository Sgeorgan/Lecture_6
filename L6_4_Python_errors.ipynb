# Introduction to Python Errors

Welcome to this Jupyter notebook designed for beginner Python programmers. Understanding and resolving errors is a crucial part of programming. Errors are not your enemies; they are helpful messages that guide you to find and fix issues in your code. In this notebook, we'll explore common types of errors you'll encounter while coding in Python and learn how to address them.

## Syntax Errors

Syntax errors occur when the Python parser detects an incorrect statement. This means there's something wrong with the way your code is written.

````python
wind_speed_km = 50
wind_speed_ms = wind_speed_km * 1000 / 3600

# On purpose: the closing quotation mark is missing, which causes a SyntaxError.
# Note that a SyntaxError stops the WHOLE cell - even the two correct lines above will not run.
print(f"A wind speed of {wind_speed_km} km/h is {wind_speed_ms} m/s.)
````

**Fixing the SyntaxError:** Ensure that all your strings are properly closed with matching quotation marks.

> ⚠️ **Common Pitfall: a `SyntaxError` often points at the *next* line**
>
> When you forget to close a bracket or parenthesis, Python keeps reading, hoping to find the closing one. It only gives up further along - so the line number it reports is often the line *after* the real mistake. (A missing closing quotation mark is usually reported on the same line, as in the example above.)
>
> If a `SyntaxError` points at a line that looks perfectly fine, check the line (or two) **above** it for an unclosed `"`, `(`, `[`, or `{`.

## Runtime Errors

Runtime errors occur while your program is running. These are not syntax errors, but rather mistakes that become apparent only when the code is executed.

### NameError example

````python
# NameError example: 'age' has never been given a value,
# so Python does not know what it refers to.
print(age)
````

**Fixing the NameError:** Make sure all variables are defined before you use them.

> ⚠️ **Common Pitfall: read the error message from the BOTTOM up**
>
> When something goes wrong, Python prints a "traceback" - often an intimidating wall of text. Don't panic, and don't start reading at the top. The useful part is at the **bottom**:
>
> - The **last line** gives the *error type* and a short message, e.g. `NameError: name 'age' is not defined`. That is usually all you need.
> - The lines **above** it show *where* it happened, with an arrow (`---->`) pointing at the line that failed.
>
> Getting into the habit of reading the last line first will save you an enormous amount of time.

### TypeError
Issue Description:
A TypeError occurs when an operation is performed on an object of an inappropriate type. This can happen when you try to concatenate a string with a non-string type without explicit conversion, or when you perform operations that are not supported by the object's type.

````python
# Attempting to concatenate a string with an integer.
# The + operator cannot join text and a number directly, so this raises a TypeError.
age = 25
message = "You are " + age + " years old."
````

**Fixing the TypeError:** Convert the number to a string so that both parts are the same type - `message = "You are " + str(age) + " years old."` - or use an f-string, which does the conversion for you: `message = f"You are {age} years old."`

### ValueError
Issue Description:
A ValueError occurs when a function receives an argument of the correct type but with an inappropriate value. This can happen when converting types or when using functions that require specific values that do not match the input.

````python
# Attempting to convert an invalid string to an integer.
# "twenty" is the right TYPE (a string) but the wrong VALUE, so this raises a ValueError.
number = int("twenty")
````

**Fixing the ValueError:** Only convert text that actually contains a number, e.g. `int("20")`. If the text comes from a user, check it first - or wrap the conversion in `try` / `except ValueError` so your program can respond politely instead of stopping.

## Logical Errors

Logical errors are the hardest to track because the code does not crash. Instead, it runs successfully but produces incorrect results.

````python
# We want the average of a morning and an evening temperature
temp_morning = 4
temp_evening = 10

# Logical error: division happens BEFORE addition, so this calculates 4 + (10 / 2) = 9.0
average_temp = temp_morning + temp_evening / 2
print(f"Average temperature: {average_temp}")
````

**Fixing the Logical Error:** Use brackets so that the addition happens first: `average_temp = (temp_morning + temp_evening) / 2`, which gives the correct answer `7.0`. Logical errors are found by checking results against values you can work out by hand.

## Handling Exceptions

Python provides a way to handle errors so that your program doesn't crash. You can anticipate potential errors and manage them using `try` and `except` blocks.

````python
# try/except lets the program carry on instead of stopping.
# Python runs the code in 'try'; if the named error happens, the 'except' block runs instead.
try:
    print(10 / 0)
except ZeroDivisionError:
    print('You cannot divide by zero.')
````

### Example: Missing colon in a control flow statement

A very common beginner mistake is forgetting the colon `:` at the end of an `if`, `for`, or `while` statement. Python then reaches the end of the line still expecting more, and reports a `SyntaxError`.

````python
# Incorrect code missing a colon
if True
    print("This statement is true.")
````

**Fixing the SyntaxError:** Add the missing colon at the end of the line, so Python knows the block is about to begin:

```python
if True:
    print("This statement is true.")
```

### Example: Indentation error

Python uses indentation to work out which lines belong inside an `if`, `for`, or function. If the indentation is missing, Python raises an `IndentationError`.

````python
# Incorrect indentation
if True:
print("This statement is true.")  # This line should be indented
````

**Fixing the IndentationError:** Indent the line that belongs inside the `if` (4 spaces is the usual choice):

```python
if True:
    print("This statement is true.")
```

#### Other kinds of errors

There are certainly [other kinds of errors and exceptions in Python](https://docs.python.org/3/tutorial/errors.html), but this list comprises those you're most likely to encounter.
As you can see, knowing the name of each error can be helpful in trying to figure out what has gone wrong, and knowing what these common error types mean will save you time trying to fix your programs.

## Summary

Errors are not failures - they are Python telling you exactly what it could not do. This lesson grouped them into three families:

- **Syntax errors** - the code is written incorrectly, so Python cannot even begin to run it.
- **Runtime errors (exceptions)** - the code is valid, but something goes wrong while it is running.
- **Logical errors** - the code runs without complaint but produces the wrong answer. These are the hardest to spot, because Python cannot help you find them.

You also met `try` / `except`, which lets your program respond to an error instead of stopping.

### Error reference

Here are the errors you have met across the whole course, gathered in one place:

| Error | What it means | Typical cause |
| --- | --- | --- |
| `SyntaxError` | Python cannot understand the code at all | Missing `:`, or an unclosed quote or bracket |
| `IndentationError` | The indentation is wrong or inconsistent | Body of an `if`/`for`/function not indented; mixing tabs and spaces |
| `NameError` | A name is not defined | Typo in a variable name, or using it before defining it |
| `TypeError` | An operation on the wrong *type* | `"text" + 5` without converting first |
| `ValueError` | Right type, but an unsuitable *value* | `int("twenty")` |
| `IndexError` | That list position doesn't exist | Using an index past the end of a list |
| `FileNotFoundError` | The file cannot be found | Wrong file name, or the file is not in the working directory |
| `ModuleNotFoundError` | The module cannot be imported | Package not installed, or your `.py` file is not in the working directory |
| `KeyError` | That dictionary key doesn't exist | Asking for a key that was never added |
| `AttributeError` | That type has no such method | Calling `.append()` on an integer |
| `ZeroDivisionError` | Division by zero | Dividing by a variable that happens to be `0` |
| `AssertionError` | An `assert` check failed | A value fell outside the range you required |

**A quick strategy when you hit an error:** read the **last line** first to get the error type, note the line number it points at, then look that error type up in the table above.
